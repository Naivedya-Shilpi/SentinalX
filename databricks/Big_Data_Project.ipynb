```python
# 1. Load UNSW-NB15 (Parquet format)
df_unsw_train = spark.read.parquet("/Volumes/workspace/default/network_traffic_data/UNSW_NB15_training-set.parquet")
df_unsw_test  = spark.read.parquet("/Volumes/workspace/default/network_traffic_data/UNSW_NB15_testing-set.parquet")

# 2. Define NSL-KDD column schema (NSL-KDD .txt files do not contain header rows)
kdd_columns = [
    "duration", "protocol_type", "service", "flag", "src_bytes", "dst_bytes",
    "land", "wrong_fragment", "urgent", "hot", "num_failed_logins", "logged_in",
    "num_compromised", "root_shell", "su_attempted", "num_root", "num_file_creations",
    "num_shells", "num_access_files", "num_outbound_cmds", "is_host_login",
    "is_guest_login", "count", "srv_count", "serror_rate", "srv_serror_rate",
    "rerror_rate", "srv_rerror_rate", "same_srv_rate", "diff_srv_rate",
    "srv_diff_host_rate", "dst_host_count", "dst_host_srv_count",
    "dst_host_same_srv_rate", "dst_host_diff_srv_rate", "dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate", "dst_host_serror_rate", "dst_host_srv_serror_rate",
    "dst_host_rerror_rate", "dst_host_srv_rerror_rate", "label", "difficulty_level"
]

# 3. Load NSL-KDD (CSV format without header)
df_kdd_train = (
    spark.read.format("csv")
    .option("inferSchema", "true")
    .load("/Volumes/workspace/default/network_traffic_data/KDDTrain+.txt")
    .toDF(*kdd_columns)
)

# Verify loads
display(df_unsw_train.limit(5))
display(df_kdd_train.limit(5))
```

```python
from pyspark.sql.functions import col, when
from pyspark.ml import Pipeline
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler, MinMaxScaler

# ---------------------------------------------------------
# 1. Standardize Target Labels
# ---------------------------------------------------------
# UNSW-NB15: already has 'label' (0 = normal, 1 = attack)
# Ensure it is cast to double/integer for MLlib
unsw_train_clean = df_unsw_train.withColumn("label", col("label").cast("double"))
unsw_test_clean  = df_unsw_test.withColumn("label", col("label").cast("double"))

# NSL-KDD: map 'normal' -> 0.0, any attack string -> 1.0
# Drop 'difficulty_level' as it's not a traffic feature
kdd_train_clean = (
    df_kdd_train
    .withColumn("label", when(col("label") == "normal", 0.0).otherwise(1.0))
    .drop("difficulty_level")
)

# ---------------------------------------------------------
# 2. Define Feature Columns for UNSW-NB15
# ---------------------------------------------------------
unsw_cat_cols = ["proto", "service", "state"]
# Exclude identifiers, text attack categories, and target label
unsw_exclude = set(unsw_cat_cols + ["id", "attack_cat", "label"])
unsw_num_cols = [c for c, dtype in unsw_train_clean.dtypes if c not in unsw_exclude and dtype in ["int", "bigint", "double", "float"]]

# ---------------------------------------------------------
# 3. Build Preprocessing Pipeline (UNSW-NB15)
# ---------------------------------------------------------
# Categorical encoding
indexers = [
    StringIndexer(inputCol=c, outputCol=f"{c}_idx", handleInvalid="keep")
    for c in unsw_cat_cols
]
encoders = [
    OneHotEncoder(inputCol=f"{c}_idx", outputCol=f"{c}_vec")
    for c in unsw_cat_cols
]

# Assemble & Scale numerical features
num_assembler = VectorAssembler(inputCols=unsw_num_cols, outputCol="num_raw", handleInvalid="keep")
# MinMaxScaler scales to [0, 1] — guarantees non-negative features for NaiveBayes
scaler = MinMaxScaler(inputCol="num_raw", outputCol="num_scaled")

# Final combined feature vector
final_assembler = VectorAssembler(
    inputCols=["num_scaled"] + [f"{c}_vec" for c in unsw_cat_cols],
    outputCol="features",
    handleInvalid="keep"
)

pipeline_unsw = Pipeline(stages=indexers + encoders + [num_assembler, scaler, final_assembler])

# Fit on training data and transform both train & test sets
print("Fitting pipeline on UNSW-NB15...")
pipeline_model_unsw = pipeline_unsw.fit(unsw_train_clean)
unsw_train_prep = pipeline_model_unsw.transform(unsw_train_clean).select("features", "label")
unsw_test_prep  = pipeline_model_unsw.transform(unsw_test_clean).select("features", "label")

# CACHING REMOVED FOR SERVERLESS COMPUTE COMPATIBILITY

print(f"UNSW Train Count: {unsw_train_prep.count()}")
print(f"UNSW Test Count: {unsw_test_prep.count()}")
display(unsw_train_prep.limit(3))
```

```python
from pyspark.sql.functions import col, when
from pyspark.ml import Pipeline
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler, MinMaxScaler

# 1. Load the Test Set for NSL-KDD
df_kdd_test = (
    spark.read.format("csv")
    .option("inferSchema", "true")
    .load("/Volumes/workspace/default/network_traffic_data/KDDTest+.txt")
    .toDF(*kdd_columns) # Reusing the schema array we defined earlier
)

# 2. Standardize Target Labels (Normal = 0.0, Attack = 1.0)
kdd_test_clean = (
    df_kdd_test
    .withColumn("label", when(col("label") == "normal", 0.0).otherwise(1.0))
    .drop("difficulty_level")
)

# 3. Define Feature Columns for NSL-KDD
kdd_cat_cols = ["protocol_type", "service", "flag"]
kdd_exclude = set(kdd_cat_cols + ["label"])
kdd_num_cols = [c for c, dtype in kdd_train_clean.dtypes if c not in kdd_exclude and dtype in ["int", "bigint", "double", "float"]]

# 4. Build Preprocessing Pipeline (NSL-KDD)
kdd_indexers = [
    StringIndexer(inputCol=c, outputCol=f"{c}_idx", handleInvalid="keep")
    for c in kdd_cat_cols
]
kdd_encoders = [
    OneHotEncoder(inputCol=f"{c}_idx", outputCol=f"{c}_vec")
    for c in kdd_cat_cols
]

kdd_num_assembler = VectorAssembler(inputCols=kdd_num_cols, outputCol="num_raw", handleInvalid="keep")
kdd_scaler = MinMaxScaler(inputCol="num_raw", outputCol="num_scaled")

kdd_final_assembler = VectorAssembler(
    inputCols=["num_scaled"] + [f"{c}_vec" for c in kdd_cat_cols],
    outputCol="features",
    handleInvalid="keep"
)

pipeline_kdd = Pipeline(stages=kdd_indexers + kdd_encoders + [kdd_num_assembler, kdd_scaler, kdd_final_assembler])

# 5. Fit and Transform
print("Fitting pipeline on NSL-KDD...")
pipeline_model_kdd = pipeline_kdd.fit(kdd_train_clean)

kdd_train_prep = pipeline_model_kdd.transform(kdd_train_clean).select("features", "label")
kdd_test_prep  = pipeline_model_kdd.transform(kdd_test_clean).select("features", "label")

print(f"KDD Train Count: {kdd_train_prep.count()}")
print(f"KDD Test Count: {kdd_test_prep.count()}")
display(kdd_train_prep.limit(3))
```

```python
from pyspark.ml.classification import (
    LogisticRegression,
    RandomForestClassifier,
    LinearSVC,
    GBTClassifier,
    NaiveBayes
)
from pyspark.ml.evaluation import (
    MulticlassClassificationEvaluator,
    BinaryClassificationEvaluator
)
from pyspark.sql.types import (
    StructType, StructField, StringType, DoubleType
)

eval_multi = MulticlassClassificationEvaluator(
    labelCol="label",
    predictionCol="prediction"
)

eval_auc = BinaryClassificationEvaluator(
    labelCol="label",
    rawPredictionCol="rawPrediction",
    metricName="areaUnderROC"
)

def get_fresh_models():
    return {
        "Logistic Regression": LogisticRegression(
            featuresCol="features", labelCol="label", maxIter=20
        ),
        "Random Forest": RandomForestClassifier(
            featuresCol="features", labelCol="label",
            numTrees=25, maxDepth=6, seed=42
        ),
        "Linear SVC": LinearSVC(
            featuresCol="features", labelCol="label", maxIter=20
        ),
        "Gradient-Boosted Trees": GBTClassifier(
            featuresCol="features", labelCol="label",
            maxIter=20, maxDepth=5, seed=42
        ),
        "Naive Bayes": NaiveBayes(
            featuresCol="features", labelCol="label",
            modelType="multinomial"
        )
    }

benchmarks = []

datasets = [
    ("UNSW-NB15", unsw_train_prep, unsw_test_prep),
    ("NSL-KDD", kdd_train_prep, kdd_test_prep)
]

for d_name, train_data, test_data in datasets:
    print(f"\n{'=' * 55}")
    print(f"TRAINING ON DATASET: {d_name}")
    print(f"{'=' * 55}")

    for name, algo in get_fresh_models().items():
        print(f"\nTraining {name} on {d_name}...")

        try:
            model = algo.fit(train_data)
            preds = model.transform(test_data)

            acc = eval_multi.evaluate(
                preds, {eval_multi.metricName: "accuracy"}
            )
            weighted_prec = eval_multi.evaluate(
                preds, {eval_multi.metricName: "weightedPrecision"}
            )
            weighted_rec = eval_multi.evaluate(
                preds, {eval_multi.metricName: "weightedRecall"}
            )
            weighted_f1 = eval_multi.evaluate(
                preds, {eval_multi.metricName: "f1"}
            )

            attack_prec = eval_multi.evaluate(
                preds,
                {
                    eval_multi.metricName: "precisionByLabel",
                    eval_multi.metricLabel: 1.0
                }
            )
            attack_rec = eval_multi.evaluate(
                preds,
                {
                    eval_multi.metricName: "recallByLabel",
                    eval_multi.metricLabel: 1.0
                }
            )
            attack_f1 = eval_multi.evaluate(
                preds,
                {
                    eval_multi.metricName: "fMeasureByLabel",
                    eval_multi.metricLabel: 1.0
                }
            )

            try:
                auc = eval_auc.evaluate(preds)
            except Exception as e:
                print(f"AUC unavailable: {str(e)[:250]}")
                auc = float("nan")

            values = [
                acc, weighted_prec, weighted_rec, weighted_f1,
                attack_prec, attack_rec, attack_f1, auc
            ]

            benchmarks.append((d_name, name, *[
                round(float(v), 4) for v in values
            ]))

            print(
                f"Accuracy={acc:.4f} | "
                f"Weighted F1={weighted_f1:.4f} | "
                f"Attack Precision={attack_prec:.4f} | "
                f"Attack Recall={attack_rec:.4f} | "
                f"Attack F1={attack_f1:.4f} | "
                f"AUC={auc:.4f}"
            )

            print("Confusion matrix counts (actual label, predicted label):")
            (
                preds.groupBy("label", "prediction")
                .count()
                .orderBy("label", "prediction")
                .show()
            )

        except Exception as e:
            print(f"FAILED: {name} on {d_name}")
            print(str(e)[:1000])

            benchmarks.append((
                d_name, name,
                float("nan"), float("nan"),
                float("nan"), float("nan"),
                float("nan"), float("nan"),
                float("nan"), float("nan")
            ))

results_schema = StructType([
    StructField("Dataset", StringType(), False),
    StructField("Model", StringType(), False),
    StructField("Accuracy", DoubleType(), True),
    StructField("Weighted_Precision", DoubleType(), True),
    StructField("Weighted_Recall", DoubleType(), True),
    StructField("Weighted_F1", DoubleType(), True),
    StructField("Attack_Precision", DoubleType(), True),
    StructField("Attack_Recall", DoubleType(), True),
    StructField("Attack_F1", DoubleType(), True),
    StructField("AUC_ROC", DoubleType(), True)
])

df_results = spark.createDataFrame(benchmarks, results_schema)

display(
    df_results.orderBy("Dataset", "Attack_F1", ascending=[True, False])
)
```

```python
from pyspark.ml.classification import NaiveBayes
from pyspark.ml.linalg import SparseVector, VectorUDT, Vectors
from pyspark.sql.functions import udf

def clamp_nonnegative(v):
    if v is None:
        return None

    if isinstance(v, SparseVector):
        indices = []
        values = []

        for i, x in zip(v.indices, v.values):
            x = max(0.0, float(x))
            if x != 0.0:
                indices.append(int(i))
                values.append(x)

        return Vectors.sparse(v.size, indices, values)

    return Vectors.dense([
        max(0.0, float(x)) for x in v
    ])

clamp_features = udf(clamp_nonnegative, VectorUDT())

unsw_train_nb = unsw_train_prep.withColumn(
    "features", clamp_features("features")
)

unsw_test_nb = unsw_test_prep.withColumn(
    "features", clamp_features("features")
)

nb_model_unsw = NaiveBayes(
    featuresCol="features",
    labelCol="label",
    modelType="multinomial"
).fit(unsw_train_nb)

nb_preds_unsw = nb_model_unsw.transform(unsw_test_nb)

for metric in ["accuracy", "weightedPrecision", "weightedRecall", "f1"]:
    score = eval_multi.evaluate(
        nb_preds_unsw,
        {eval_multi.metricName: metric}
    )
    print(f"{metric}: {score:.4f}")

print("AUC-ROC:")
try:
    print(round(eval_auc.evaluate(nb_preds_unsw), 4))
except Exception as e:
    print("AUC evaluation failed:", str(e)[:500])

nb_preds_unsw.groupBy("label", "prediction").count().orderBy(
    "label", "prediction"
).show()
```

```python

from pyspark.ml.classification import NaiveBayes
from pyspark.ml.evaluation import MulticlassClassificationEvaluator, BinaryClassificationEvaluator
from pyspark.ml.linalg import SparseVector, VectorUDT, Vectors
from pyspark.sql.functions import udf, col
from pyspark.sql.types import DoubleType
from pyspark.sql import functions as F

def clamp_nonnegative(v):
    if v is None:
        return None
    if isinstance(v, SparseVector):
        indices = []
        values = []
        for i, x in zip(v.indices, v.values):
            x = max(0.0, float(x))
            if x != 0.0:
                indices.append(int(i))
                values.append(x)
        return Vectors.sparse(v.size, indices, values)
    return Vectors.dense([max(0.0, float(x)) for x in v])

clamp_features = udf(clamp_nonnegative, VectorUDT())

unsw_train_nb = unsw_train_prep.withColumn("features", clamp_features("features"))
unsw_test_nb = unsw_test_prep.withColumn("features", clamp_features("features"))

nb_model_unsw = NaiveBayes(
    featuresCol="features",
    labelCol="label",
    modelType="multinomial"
).fit(unsw_train_nb)

nb_preds_unsw = nb_model_unsw.transform(unsw_test_nb)

eval_multi = MulticlassClassificationEvaluator(
    labelCol="label",
    predictionCol="prediction"
)

eval_auc = BinaryClassificationEvaluator(
    labelCol="label",
    rawPredictionCol="rawPrediction",
    metricName="areaUnderROC"
)

metrics = {}

for metric in ["accuracy", "weightedPrecision", "weightedRecall", "f1"]:
    metrics[metric] = eval_multi.evaluate(
        nb_preds_unsw,
        {eval_multi.metricName: metric}
    )

metrics["attack_precision"] = eval_multi.evaluate(
    nb_preds_unsw,
    {eval_multi.metricName: "precisionByLabel", eval_multi.metricLabel: 1.0}
)

metrics["attack_recall"] = eval_multi.evaluate(
    nb_preds_unsw,
    {eval_multi.metricName: "recallByLabel", eval_multi.metricLabel: 1.0}
)

metrics["attack_f1"] = eval_multi.evaluate(
    nb_preds_unsw,
    {eval_multi.metricName: "fMeasureByLabel", eval_multi.metricLabel: 1.0}
)

metrics["auc_roc"] = eval_auc.evaluate(nb_preds_unsw)

for name, value in metrics.items():
    print(f"{name}: {value:.4f}")

export_path = "/Volumes/workspace/default/network_traffic_data/final_benchmarks_csv/"

old_results = spark.read.option("header", True).option("inferSchema", True).csv(export_path)

corrected_row = spark.createDataFrame(
    [(
        "UNSW-NB15",
        "Naive Bayes",
        round(metrics["accuracy"], 4),
        round(metrics["weightedPrecision"], 4),
        round(metrics["weightedRecall"], 4),
        round(metrics["f1"], 4),
        round(metrics["attack_precision"], 4),
        round(metrics["attack_recall"], 4),
        round(metrics["attack_f1"], 4),
        round(metrics["auc_roc"], 4)
    )],
    schema=old_results.schema
)

corrected_results = (
    old_results
    .filter(
        ~(
            (col("Dataset") == "UNSW-NB15") &
            (col("Model") == "Naive Bayes")
        )
    )
    .unionByName(corrected_row)
)

display(corrected_results.orderBy("Dataset", "Model"))
```

```python
from pyspark.sql import functions as F

nb_counts = nb_preds_unsw.agg(
    F.sum(F.when((F.col("label") == 0) & (F.col("prediction") == 0), 1).otherwise(0)).alias("TN"),
    F.sum(F.when((F.col("label") == 0) & (F.col("prediction") == 1), 1).otherwise(0)).alias("FP"),
    F.sum(F.when((F.col("label") == 1) & (F.col("prediction") == 0), 1).otherwise(0)).alias("FN"),
    F.sum(F.when((F.col("label") == 1) & (F.col("prediction") == 1), 1).otherwise(0)).alias("TP")
)

confusion_results = (
    nb_counts
    .withColumn("Dataset", F.lit("UNSW-NB15"))
    .withColumn("Model", F.lit("Naive Bayes"))
    .select("Dataset", "Model", "TN", "FP", "FN", "TP")
)

display(confusion_results)
```
